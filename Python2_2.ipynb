```python
#H1 Write a function to calculate the sum of numbers from 1 to N.

#_____________________SOLUTION:________________________________

def sum_of_numbers(N):
    total = 0
    for i in range(1, N+1):
        total += i
    return total
N = int(input("Enter the value of N: "))
print("The sum of numbers from 1 to", N, "is", sum_of_numbers(N))
```

```python
#H7 Write a Python program to calculate the mean of a list of numbers.
#____________________SOLUTION______________________________________:

import math
def calculate_mean(numbers):
    if len(numbers) == 0:
        return None  # Handle empty list safely
    else:
      return sum(numbers) / len(numbers)

# Example usage
data = [10, 15, 20, 25, 30, 35, 40, 45, 50, 55]
print("Mean of the list:", calculate_mean(data))

```

```python
#H8 Classify Numbers into Even and Odd Lists
#____________________SOLUTION______________________________________:

def classify_numbers(numbers):
    even_list = []
    odd_list = []

    for n in numbers:
        if n % 2 == 0:
            even_list.append(n)
        else:
            odd_list.append(n)

    return even_list, odd_list


data = [10, 15, 20, 25, 30, 35, 40, 45, 50]

evens, odds = classify_numbers(data)

print("Even numbers:", evens)
print("Odd numbers:", odds)

```

```python
## String Functions

# Initial string
s = "Skill Morph"

# 1. upper()
print(s.upper())  # Converts all characters to uppercase
# 2. lower()
print(s.lower())  # Converts all characters to lowercase
# 3. partition()
print(s.partition(' '))  # Splits the string at the first occurrence of ' '
# 4. replace()
print(s.replace('Morph', 'Master'))  # Replaces 'Morph' with 'Master'
# 5. find()
print(s.find('Morph'))  # Returns the index of 'Morph' in the string, or -1 if not found
# 6. rstrip()
print(s.rstrip())  # Removes trailing spaces from the string
# 7. split()
print(s.split())  # Splits the string into a list at spaces
# 8. startswith()
print(s.startswith('Skill'))  # Checks if the string starts with 'Skill'
# 9. isnumeric()
print(s.isnumeric())  # Checks if all characters in the string are numeric (returns False)
# 10. index()
print(s.index('Morph'))  # Returns the index of the first occurrence of 'Morph'
```

```python
#____________________SOLUTION____OF   H_9 __________________________________:
def max_value_key(d):
    # max() with key argument finds the dictionary key with the largest value
    return max(d, key=d.get)

data = {'a': 10, 'b': 20, 'c': 15}
print("Key with maximum value:", max_value_key(data))
```

```python
#____________________SOLUTION____OF   H_10  __________________________________:
def squares_dict(n):

    return {i: i**2 for i in range(1, n+1)}

n = 5
print("Dictionary of squares:", squares_dict(n))

```
